**Studi Kasus 1**

Pada studi kasus ini, kita melakukan klasisfikasi suhu berdasarkan prinsip Marry yaitu, panas dan dingin.

Persepsi ini ditentukan oleh  dua varibel yaitu, Temperatur Udara dan kecepatan Angin.

Metode klasifikasi yang digunakan adalah KNN (K_Nearest Neighbors)

**Pengerjaannya**

```python
# --- Import Library ---
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# --- Dataset (sesuai tabel pada soal) ---
data = {
    "Temperatur": [10, 25, 15, 20, 18, 20, 22, 24],
    "Angin":      [0, 0, 5, 3, 7, 10, 5, 6],
    "Label":      ["Dingin", "Panas", "Dingin", "Panas", "Dingin", "Dingin", "Panas", "Panas"]
}

df = pd.DataFrame(data)
df
```

**Pre Processing + Split Data**

```python
# Memisahkan fitur dan label
X = df[["Temperatur", "Angin"]]
y = df["Label"]

# Split dataset untuk mencari K terbaik
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

**Mencari Nilai K Terbaik**

```python
akurasi_k = {}

for k in range(1, len(X_train) + 1):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    akurasi = accuracy_score(y_test, y_pred)
    akurasi_k[k] = akurasi

akurasi_k
```

**Menentukan K dengan Akurasi Tertinggi**

```python
k_terbaik = max(akurasi_k, key=akurasi_k.get)
k_terbaik
```

**Melatih Model dengan K Terbaik**

```python
model_final = KNeighborsClassifier(n_neighbors=k_terbaik)
model_final.fit(X, y)
```

**Prediksi Marry**

```python
sample_df = pd.DataFrame([[16, 3]], columns=X.columns)
hasil = model_final.predict(sample_df)
hasil[0]
```

```python
# --------------------------------------------
# LATIHAN 3 - PREDIKSI CUACA MENGGUNAKAN KNN
# --------------------------------------------

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score

# --------------------------------------------------------
# 1. MEMASUKKAN DATA SECARA MANUAL SESUAI TABEL PADA GAMBAR
# --------------------------------------------------------
data = {
    "Temperature": [14.0, 39.0, 30.0, 38.0, 27.0, 32.0, -2.0, 3.0, 3.0, 28.0],
    "Humidity": [73, 96, 64, 83, 74, 55, 97, 85, 83, 74],
    "Wind Speed": [9.5, 8.5, 7.0, 1.5, 17.0, 3.5, 8.0, 6.0, 6.0, 8.5],
    "Precipitation (%)": [82.0, 71.0, 16.0, 82.0, 66.0, 26.0, 86.0, 96.0, 66.0, 107.0],
    "Cloud Cover": [
        "partly cloudy", "partly cloudy", "clear", "clear", "overcast",
        "overcast", "overcast", "partly cloudy", "clear", "clear"
    ],
    "Atmospheric Pressure": [1010.82, 1011.43, 1018.72, 1026.25, 990.67, 1010.03, 990.87, 984.46, 999.44, 1012.13],
    "UV Index": [2, 7, 5, 7, 1, 2, 1, 1, 0, 8],
    "Season": ["Winter", "Spring", "Spring", "Spring", "Winter", "Summer", "Winter", "Winter", "Winter", "Winter"],
    "Visibility (km)": [3.5, 10.0, 5.5, 1.0, 2.5, 5.0, 4.0, 3.5, 1.0, 7.5],
    "Location": ["inland", "inland", "mountain", "coastal", "mountain", "inland", "inland", "inland", "inland", "coastal"],
    "Weather Type": ["Rainy", "Cloudy", "Sunny", "Sunny", "Rainy", "Cloudy", "Snowy", "Snowy", "Snowy", "Sunny"]
}

df = pd.DataFrame(data)

print("DATASET:")
display(df)

# -------------------------------
# 2. LABEL ENCODING UNTUK FITUR
# -------------------------------

le_cloud = LabelEncoder()
le_season = LabelEncoder()
le_location = LabelEncoder()
le_weather = LabelEncoder()

df["Cloud Cover"] = le_cloud.fit_transform(df["Cloud Cover"])
df["Season"] = le_season.fit_transform(df["Season"])
df["Location"] = le_location.fit_transform(df["Location"])
df["Weather Type"] = le_weather.fit_transform(df["Weather Type"])

# -------------------------------
# 3. SPLIT DATA
# -------------------------------
X = df.drop("Weather Type", axis=1)
y = df["Weather Type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# -------------------------------
# 4. TRAIN KNN
# -------------------------------

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# -------------------------------
# 5. PREDIKSI & EVALUASI
# -------------------------------

y_pred = knn.predict(X_test)

print("\nAkurasi Model KNN :", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# -------------------------------
# 6. CONTOH PREDIKSI MANUAL
# -------------------------------
contoh = [[20, 80, 5, 50, 1, 1005, 3, 2, 4, 0]]  # Data acak contoh
hasil = knn.predict(contoh)
print("\nContoh Prediksi Cuaca:", le_weather.inverse_transform(hasil)[0])
```
