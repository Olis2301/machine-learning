```python
import pandas as pd
df = pd.read_csv('winemag-data-130k-v2.csv')
df.head()
```

```python
df.info()
df.describe()

```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include='number')
df[num_cols.columns] = df[num_cols.columns].fillna(num_cols.mean())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df.iloc[:,0], bins=30, kde=True)
plt.title("Distribusi Data Kolom Pertama")
plt.show()

```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='price', y='points', data=df)
plt.title('Hubungan Harga Wine vs Rating')
plt.xlabel('Price')
plt.ylabel('Points')
plt.show()
```

```python
top_countries = df['country'].value_counts().head(5).index
plt.figure(figsize=(10,6))
sns.boxplot(x='country', y='price', data=df[df['country'].isin(top_countries)])
plt.title('Distribusi Harga Wine (Top 5 Negara)')
plt.xticks(rotation=45)
plt.show()
```
